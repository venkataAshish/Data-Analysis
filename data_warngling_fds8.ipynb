```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Amit', 'Ravi', 'Priya'],
    'Marks': [85, 90, 88]
})

df2 = pd.DataFrame({
    'ID': [4, 5, 6],
    'Name': ['John', 'Sara', 'Raj'],
    'Marks': [78, 92, 80]
})

print(df1)
print(df2)
```

```python
df = pd.DataFrame({
    'Department': ['IT', 'IT', 'CSE', 'CSE'],
    'Student': ['Amit', 'Ravi', 'Priya', 'John'],
    'Marks': [85, 90, 88, 78]
})

df = df.set_index(['Department', 'Student'])

print(df)
```

```python
result = pd.concat([df1, df2], ignore_index=True)

print(result)
```

```python
df3 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Age': [20, 21, 22]
})

merged = pd.merge(df1, df3, on='ID')

print(merged)
```

```python
df = pd.DataFrame({
    'Name': ['Amit', 'Ravi'],
    'Maths': [85, 90],
    'Science': [80, 88]
})

result = pd.melt(
    df,
    id_vars=['Name'],
    var_name='Subject',
    value_name='Marks'
)

print(result)
```

```python
df = pd.DataFrame({
    'Name': ['Amit', 'Amit', 'Ravi', 'Ravi'],
    'Subject': ['Maths', 'Science', 'Maths', 'Science'],
    'Marks': [85, 80, 90, 88]
})

result = df.pivot(
    index='Name',
    columns='Subject',
    values='Marks'
)

print(result)
```

```python
result = pd.pivot_table(
    df,
    index='Name',
    columns='Subject',
    values='Marks',
    aggfunc='mean'
)

print(result)
```

```python
# Convert columns into an additional index level
stacked = result.stack()
print(stacked)

# Convert an index level back into columns
unstacked = stacked.unstack()
print(unstacked)
```
