```python
import pandas as pd
```

```python
emp1 = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Sneha", "Arjun", "Priya", "Kiran"],
    "Dept": ["IT", "HR", "Finance", "IT", "Marketing"]
})

print("EMPLOYEE DATAFRAME 1 ")
print(emp1)


emp2 = pd.DataFrame({
    "Emp_ID": [106, 107, 108, 109, 110],
    "Name": ["Anil", "Kavya", "Rohit", "Meena", "Suresh"],
    "Dept": ["HR", "IT", "Finance", "Marketing", "IT"]
})

print("\nEMPLOYEE DATAFRAME 2")
print(emp2)
```

```python
salary = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105,
               106, 107, 108, 109, 110],
    "Salary": [45000, 40000, 55000, 50000, 42000,
               48000, 46000, 58000, 44000, 52000]
})

print("\n===== SALARY DATAFRAME =====")
print(salary)
```

```python
emp = pd.concat([emp1, emp2], ignore_index=True)
print("\\nCONCATENATION")
print(emp)
```

```python
df1 = pd.DataFrame({
    "Name": ["Rahul", "Anil", "Kavya"],
    "Age": [20, 21, 20]
})

df2 = pd.DataFrame({
    "Marks": [85, 78, 92],
    "Grade": ["A", "B", "A"]
})

concat_columns = pd.concat([df1, df2], axis=1)
print("\n COLUMN-WISE CONCATENATION")
print(concat_columns)
```

```python
inner_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="inner"
)

print("\n===== INNER MERGE =====")
print(inner_merge)
```

```python
left_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="left"
)

print("\n===== LEFT MERGE =====")
print(left_merge)
```

```python
right_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="right"
)

print("\n===== RIGHT MERGE =====")
print(right_merge)
```

```python
outer_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="outer"
)

print("\n===== OUTER MERGE =====")
print(outer_merge)
```

```python
employee = pd.DataFrame({
    "Employee_ID": [101, 102, 103],
    "Name": ["Rahul", "Anil", "Kavya"]
})

employee_salary = pd.DataFrame({
    "ID": [101, 102, 103],
    "Salary": [50000, 48000, 46000]
})
```

```python
merge_different_columns = pd.merge(
    employee,
    employee_salary,
    left_on="Employee_ID",
    right_on="ID"
)

print("\n===== MERGE USING DIFFERENT COLUMN NAMES =====")
print(merge_different_columns)
```

```python
emp_details = pd.DataFrame({
    "Name": ["Rahul", "Anil", "Kavya"],
    "Dept": ["IT", "HR", "Finance"]
}, index=[101, 102, 103])


emp_salary = pd.DataFrame({
    "Salary": [50000, 48000, 46000]
}, index=[101, 102, 103])


join_result = emp_details.join(emp_salary)

print("\n===== JOIN =====")
print(join_result)
```

```python
left_join = emp_details.join(
    emp_salary,
    how="left"
)

print("\n===== LEFT JOIN =====")
print(left_join)
```

```python
right_join = emp_details.join(
    emp_salary,
    how="right"
)

print("\n===== RIGHT JOIN =====")
print(right_join)
```

```python

outer_join = emp_details.join(
    emp_salary,
    how="outer"
)

print("\n===== OUTER JOIN =====")
print(outer_join)
```

```python
final_df = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="inner"
)

print("\n===== FINAL EMPLOYEE DATA =====")
print(final_df)
```
