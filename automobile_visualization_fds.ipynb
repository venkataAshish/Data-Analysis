```python
df = pd.read_csv("Automobile_data.csv")
```

```python
df.shape
```

```python
df.columns = [
    "symboling", "normalized-losses", "make", "fuel-type",
    "aspiration", "num-of-doors", "body-style", "drive-wheels",
    "engine-location", "wheel-base", "length", "width", "height",
    "curb-weight", "engine-type", "num-of-cylinders", "engine-size",
    "fuel-system", "bore", "stroke", "compression-ratio",
    "horsepower", "peak-rpm", "city-mpg", "highway-mpg", "price"
]
df.head()
```

```python
numeric_cols = [
    "symboling", "normalized-losses", "wheel-base", "length",
    "width", "height", "curb-weight", "engine-size", "bore",
    "stroke", "compression-ratio", "horsepower", "peak-rpm",
    "city-mpg", "highway-mpg", "price"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
plt.hist(df["price"].dropna(), bins=20, color="skyblue")
plt.title("Distribution of Car Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
df["make"].value_counts().plot(kind="bar")
plt.title("Cars by Manufacturer")
plt.xlabel("Manufacturer")
plt.ylabel("Count")
plt.show()
```

```python
df["horsepower"] = pd.to_numeric(df["horsepower"], errors="coerce")

plt.scatter(df["horsepower"], df["price"])
plt.title("Horsepower vs Price")
plt.xlabel("Horsepower")
plt.ylabel("Price")
plt.show()
```

```python
sns.boxplot(data=df, x="body-style", y="price")
plt.title("Car Price by Body Style")
plt.xticks(rotation=45)
plt.show()
```

```python
cols = ["price", "horsepower", "engine-size",
        "curb-weight", "highway-mpg"]

for col in cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

sns.heatmap(df[cols].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
