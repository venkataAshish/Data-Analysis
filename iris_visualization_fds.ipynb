```python
df.hist(figsize=(8, 6))
plt.show()
```

```python
plt.figure(figsize = (6,4))
sns.histplot(df['PetalLengthCm'], bins = 20, kde = True)
plt.title("Petal Distribution")
plt.show()
```

```python
plt.figure(figsize = (6,4))
sns.histplot(df['PetalWidthCm'], bins = 20, kde = True)
plt.title("Petal-width-Distribution")
plt.show()
```

```python
plt.figure(figsize = (6,4))
sns.histplot(df['SepalLengthCm'], bins = 20, kde = True)
plt.title("Sepal-Length-Distribution")
plt.show()
```

```python
plt.figure(figsize = (6,4))
sns.histplot(df['SepalWidthCm'], bins = 20, kde = True)
plt.title("Sepal-Width- Distribution")
plt.show()
```

```python
sns.scatterplot(data=df, x="SepalLengthCm", y="PetalLengthCm", hue="Species")
plt.show()
```

```python
sns.boxplot(data=df, x="Species", y="PetalLengthCm")
plt.show()
```

```python
sns.heatmap(df.select_dtypes(include="number").corr(), annot=True)
plt.show()
```
